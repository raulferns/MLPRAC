# Experiment 1: Linear Regression
**Case Study:** Predict House Prices based on Area (10-row assumed dataset)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

### Step 1: Create Assumed Dataset (10 Rows)

```python
data = {
    'Area_sqft': [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800, 3000],
    'Price_Lakhs': [42, 50, 62, 75, 88, 102, 110, 126, 138, 152]
}

df = pd.DataFrame(data)
print("--- Assumed Dataset (10 Rows) ---")
print(df)

X = df['Area_sqft']
y = df['Price_Lakhs']
```

### Step 2: Perform Linear Regression using `linregress()`

```python
slope, intercept, r_value, p_value, std_err = linregress(X, y)

# Predict values using y = m*x + c
y_pred = slope * X + intercept

print(f"Slope     : {slope:.4f}")
print(f"Intercept : {intercept:.4f}")
print(f"R-value   : {r_value:.4f}")
```

### Step 3: Calculate Accuracy Measures (R2, MSE, RMSE, MAE)

```python
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y, y_pred)

print(f"R2 Score  : {r2:.4f}")
print(f"MSE       : {mse:.4f}")
print(f"RMSE      : {rmse:.4f}")
print(f"MAE       : {mae:.4f}")

# Predict for a new custom input
new_area = 1600
predicted_price = slope * new_area + intercept
print(f"\nPredicted Price for {new_area} sqft: {predicted_price:.2f} Lakhs")
```

### Step 4: Plot Actual vs Regression Line

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, y_pred, color='red', label='Regression Line')
plt.xlabel('Area (sq ft)')
plt.ylabel('Price (in Lakhs)')
plt.title('Linear Regression: House Area vs Price')
plt.legend()
plt.grid(True)
plt.show()
```
